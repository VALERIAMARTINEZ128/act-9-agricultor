```python
# Simulador de IA para un Agricultor
# Este programa simula un asistente inteligente usando solo listas, condicionales y un ciclo while.

# Creamos la base de conocimientos de la IA (empieza vacía como se solicitó)
# Guardará las plantas o cultivos que el agricultor ha registrado como "saludables".
cultivos_saludables = []

# Variables para el control del menú e interacción
opcion_menu = ""
mensaje_bienvenida = "¡Hola! Soy AgroAI, tu asistente virtual de agricultura."

print("====================================================")
print(mensaje_bienvenida)
print("====================================================")

# Ciclo principal interactivo
while opcion_menu != "4":
    # Mostramos el menú en cada iteración
    print("\n--- MENÚ DE AGROAI ---")
    print("1. Consultar estado de un cultivo (Diagnóstico)")
    print("2. Registrar un cultivo como saludable (Aprendizaje)")
    print("3. Ver lista de cultivos bajo control")
    print("4. Salir del sistema")

    opcion_menu = input("\nSelecciona una opción (1-4): ")

    # ----------------------------------------------------
    # OPCIÓN 1: DIAGNÓSTICO INTELIGENTE (Simula comprensión de lenguaje)
    # ----------------------------------------------------
    if opcion_menu == "1":
        print("\n[AgroAI] Describe brevemente cómo se ve tu planta o cultivo.")
        print("Ejemplos de palabras clave: 'manchas', 'amarilla', 'secas', 'normal'")

        estado = input("Tu descripción: ")
        # Convertimos a minúsculas para que la IA sea más flexible al responder
        estado = estado.lower()

        # El motor de decisiones de la IA analiza las palabras clave
        if "manchas" in estado:
            print("\n🤖 [AgroAI] ALERTA: Podría ser un hongo. Recomiendo reducir el riego y aplicar un fungicida orgánico.")
        elif "amarilla" in estado:
            print("\n🤖 [AgroAI] DIAGNÓSTICO: Posible falta de nitrógeno en el suelo o exceso de agua. Revisa el drenaje.")
        elif "secas" in estado:
            print("\n🤖 [AgroAI] DIAGNÓSTICO: Tus plantas necesitan agua urgentemente o están expuestas a demasiado sol.")
        elif "normal" in estado or "bien" in estado or "verde" in estado:
            print("\n🤖 [AgroAI] EXCELENTE: Tu cultivo se encuentra en condiciones óptimas. ¡Sigue así!")
        else:
            print("\n🤖 [AgroAI] No logré identificar el problema con exactitud. Intenta usar palabras como 'manchas', 'amarilla' o 'secas'.")

    # ----------------------------------------------------
    # OPCIÓN 2: APRENDIZAJE (La IA guarda datos en la lista)
    # ----------------------------------------------------
    elif opcion_menu == "2":
        nuevo_cultivo = input("\n¿Qué cultivo quieres registrar como saludable? (ej. Maíz, Tomate): ")
        # Convertimos a minúsculas para estandarizar
        nuevo_cultivo = nuevo_cultivo.lower()

        # Validamos que no esté ya en la lista
        if nuevo_cultivo in cultivos_saludables:
            print(f"\n🤖 [AgroAI] Ya sabía que el {nuevo_cultivo} estaba saludable. ¡Monitoreo activo!")
        else:
            # La IA "aprende" añadiendo el elemento a la lista
            cultivos_saludables.append(nuevo_cultivo)
            print(f"\n🤖 [AgroAI] Entendido. He aprendido que tu cultivo de '{nuevo_cultivo}' está sano y lo he guardado.")

    # ----------------------------------------------------
    # OPCIÓN 3: MOSTRAR MEMORIA DE LA IA
    # ----------------------------------------------------
    elif opcion_menu == "3":
        print("\n--- CULTIVOS SALUDABLES REGISTRADOS ---")
        # Si la lista está vacía
        if cultivos_saludables == []:
            print("🤖 [AgroAI] Mi memoria está vacía. No has registrado cultivos sanos todavía.")
        else:
            # Mostramos la lista completa tal cual
            print(f"🤖 [AgroAI] Cultivos bajo monitoreo seguro: {cultivos_saludables}")

    # ----------------------------------------------------
    # OPCIÓN 4: SALIDA
    # ----------------------------------------------------
    elif opcion_menu == "4":
        print("\n🤖 [AgroAI] Apagando sistema de asistencia. ¡Buena cosecha, agricultor!")

    # ----------------------------------------------------
    # MANEJO DE ERRORES DEL MENÚ
    # ----------------------------------------------------
    else:
        print("\n❌ Opción no válida. Por favor, elige un número del 1 al 4.")

print("\n====================================================")
print("Programa finalizado con éxito.")
print("====================================================")
```
